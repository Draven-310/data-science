```python
import pandas as pd
import matplotlib.pyplot as plt
 as plt
import seaborn as sb
dataset=pd.read_excel("HousePricePrediction.xlsx")
print(dataset.head(10))
print('\n\n\n\n\n\n\n')
dataset.shape
```

```python
object_cols = dataset.select_dtypes(include=['object']).columns
print("categorical variables:",len(object_cols))

integer_cols = dataset.select_dtypes(include=['int64']).columns
print("Integer variables:", len(integer_cols))

float_cols = dataset.select_dtypes(include=['float64']).columns
print("Float variables:", len(float_cols))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
numerical_dataset= dataset.select_dtypes(include=['int64','float64'])

plt.figure(figsize=(12,6))
sb.heatmap(numerical_dataset.corr(),
           cmap='BrBG',
           fmt='.2f',
           linewidths=2,
           annot=True)
plt.title("Correction Heatmap of Numerical Features")
plt.tight_layout()
plt.savefig("Correlation_Heatup.png")
print("Heatmap Saves as correlation_heatmap.png")







```

```python
unique_val=[]
for col in object_cols:
  unique_val.append(dataset[col].unique().size)
plt.figure(figsize=(10,6))
plt.title('No. unique values of Categorical Features')
plt.xticks(rotation=90)
sb.barplot(x=object_cols,y=unique_val)
```
