```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
bnb=pd.read_csv('Airbnb_Open_Data.csv')
bnb.head(10)
#bnb.shape
```

```python

print(bnb.describe())
print(bnb.isnull().sum())
print(bnb.info())
```

```python
print(bnb.columns)
```

```python
print(bnb.columns.tolist())
```

```python
bnb.drop_duplicates(inplace=True)
bnb.info()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(bnb['price'], bins=50, kde=True)
plt.title('Distribution of Listing Prices')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='room type', data=bnb , color='turquoise')
plt.title('Room Type Distribution')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.show()
```
